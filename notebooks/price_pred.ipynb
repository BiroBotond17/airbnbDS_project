```python
# Revenue impact simulation
# -------------------------
# This function evaluates the business impact of the pricing model by comparing
# current listing prices with model-recommended prices.
#
# Main logic:
# 1. Predict an adjusted "recommended" price for each listing using the trained model.
# 2. Limit price changes with upper/lower caps (max_up, max_down) so recommendations stay realistic.
# 3. Estimate current occupancy using a proxy based on Availability 365:
#       occupied_nights_proxy = 365 - Availability 365
# 4. Adjust expected occupied nights using a simple price elasticity assumption:
#       higher prices -> slightly lower occupancy
#       lower prices -> slightly higher occupancy
# 5. Calculate current and recommended revenue proxies:
#       revenue = price × occupied nights proxy
# 6. Aggregate the results into summary metrics such as:
#       average price change, % of listings increased/decreased,
#       total revenue proxy, and estimated revenue uplift.
#
# Important note:
# This is a hypothetical scenario analysis, not a causal estimate of real revenue.
# It uses availability as a proxy for occupancy and assumes a simplified relationship
# between price changes and demand.
```

```python
import numpy as np
import pandas as pd

def simulate_revenue_impact(
    df,
    features,
    model,
    take_rate=1.0,
    max_up=0.15,
    max_down=0.15,
    elasticity=0.5,
    apply_downward=True,
    return_detail=False
):
    eval_cols = features + ["Price", "Availability 365", "log_price"]
    sim = df[eval_cols].dropna().copy()

    pred_log_price = model.predict(sim[features])
    pred_price_raw = np.expm1(pred_log_price)
    current_price = sim["Price"].to_numpy()
    availability = sim["Availability 365"].to_numpy()

    lower_bound = current_price * (1 - max_down)
    upper_bound = current_price * (1 + max_up)

    recommended_price = np.clip(pred_price_raw, lower_bound, upper_bound)

    if not apply_downward:
        recommended_price = np.maximum(recommended_price, current_price)

    price_change_pct = (recommended_price - current_price) / current_price

    occupied_nights_proxy = np.clip(365 - availability, 0, 365)

    occ_multiplier = 1 - elasticity * price_change_pct
    recommended_nights_proxy = np.clip(occupied_nights_proxy * occ_multiplier, 0, 365)

    current_revenue_proxy = current_price * occupied_nights_proxy
    recommended_revenue_proxy = recommended_price * recommended_nights_proxy

    current_company_revenue_proxy = current_revenue_proxy * take_rate
    recommended_company_revenue_proxy = recommended_revenue_proxy * take_rate

    summary = pd.Series({
        "listings_evaluated": len(sim),
        "avg_current_price": current_price.mean(),
        "avg_recommended_price": recommended_price.mean(),
        "avg_price_change_pct": price_change_pct.mean(),
        "pct_listings_increased": (price_change_pct > 0).mean(),
        "pct_listings_decreased": (price_change_pct < 0).mean(),
        "current_revenue_proxy_total": current_revenue_proxy.sum(),
        "recommended_revenue_proxy_total": recommended_revenue_proxy.sum(),
        "revenue_uplift_proxy_total": recommended_revenue_proxy.sum() - current_revenue_proxy.sum(),
        "revenue_uplift_proxy_pct": recommended_revenue_proxy.sum() / current_revenue_proxy.sum() - 1,
        "current_company_revenue_proxy_total": current_company_revenue_proxy.sum(),
        "recommended_company_revenue_proxy_total": recommended_company_revenue_proxy.sum(),
        "company_revenue_uplift_proxy_total": recommended_company_revenue_proxy.sum() - current_company_revenue_proxy.sum(),
        "company_revenue_uplift_proxy_pct": recommended_company_revenue_proxy.sum() / current_company_revenue_proxy.sum() - 1
    })

    if return_detail:
        detail = sim.copy()
        detail["pred_log_price"] = pred_log_price
        detail["pred_price_raw"] = pred_price_raw
        detail["recommended_price"] = recommended_price
        detail["price_change_pct"] = price_change_pct
        detail["occupied_nights_proxy"] = occupied_nights_proxy
        detail["recommended_nights_proxy"] = recommended_nights_proxy
        detail["current_revenue_proxy"] = current_revenue_proxy
        detail["recommended_revenue_proxy"] = recommended_revenue_proxy
        return detail, summary

    return summary
```

```python
import pandas as pd
from itertools import product

results = []

max_up_values = [0.05, 0.10, 0.15, 0.20]
max_down_values = [0.00, 0.05, 0.10]
elasticity_values = [0.1, 0.2, 0.3, 0.5]
apply_downward_values = [False, True]

for max_up, max_down, elasticity, apply_downward in product(
    max_up_values,
    max_down_values,
    elasticity_values,
    apply_downward_values
):
    summary = simulate_revenue_impact(
        df_small,
        features,
        xgb_model,
        take_rate=1.0,
        max_up=max_up,
        max_down=max_down,
        elasticity=elasticity,
        apply_downward=apply_downward,
        return_detail=False
    )

    result = summary.to_dict()
    result["max_up"] = max_up
    result["max_down"] = max_down
    result["elasticity"] = elasticity
    result["apply_downward"] = apply_downward

    results.append(result)

scenario_results = pd.DataFrame(results)
```

```python
import time
import numpy as np
import pandas as pd

mc_results = []

n_test = 5

start = time.time()

for _ in range(n_test):
    elasticity = np.random.uniform(0.1, 0.4)
    max_up = np.random.uniform(0.05, 0.15)
    max_down = np.random.uniform(0.05, 0.15)

    summary = simulate_revenue_impact(
        df_small,
        features,
        xgb_model,
        take_rate=1.0,
        max_up=max_up,
        max_down=max_down,
        elasticity=elasticity,
        apply_downward=True,
        return_detail=False
    )

    row = summary.to_dict()
    row["elasticity"] = elasticity
    row["max_up"] = max_up
    row["max_down"] = max_down
    mc_results.append(row)

end = time.time()

elapsed = end - start
avg_per_run = elapsed / n_test

print(f"Total time for {n_test} runs: {elapsed:.2f} seconds")
print(f"Average time per run: {avg_per_run:.2f} seconds")
print(f"Estimated time for 100 runs: {avg_per_run * 100 / 60:.2f} minutes")
```

```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Estimated runtime settings
avg_per_run = 4.10
target_minutes = 30
n_sim = int((target_minutes * 60) / avg_per_run)

print(f"Running approximately {n_sim} simulations (~{target_minutes} minutes)...")

mc_results = []
start = time.time()

for i in range(n_sim):
    elasticity = np.random.uniform(0.1, 0.4)
    max_up = np.random.uniform(0.05, 0.15)
    max_down = np.random.uniform(0.05, 0.15)

    summary = simulate_revenue_impact(
        df_small,
        features,
        xgb_model,
        take_rate=1.0,
        max_up=max_up,
        max_down=max_down,
        elasticity=elasticity,
        apply_downward=True,
        return_detail=False
    )

    row = summary.to_dict()
    row["elasticity"] = elasticity
    row["max_up"] = max_up
    row["max_down"] = max_down
    mc_results.append(row)

    if (i + 1) % 25 == 0:
        elapsed = time.time() - start
        print(f"{i+1}/{n_sim} done | elapsed: {elapsed/60:.1f} min")

mc_df = pd.DataFrame(mc_results)

total_elapsed = time.time() - start
print(f"\nFinished in {total_elapsed/60:.2f} minutes")

# Summary stats
print("\nRevenue uplift summary:")
print(mc_df["revenue_uplift_proxy_pct"].describe())

# Best scenarios
best_runs = mc_df.sort_values("revenue_uplift_proxy_pct", ascending=False).head(10)
print("\nTop 10 scenarios:")
display(best_runs[[
    "max_up", "max_down", "elasticity",
    "avg_price_change_pct",
    "revenue_uplift_proxy_total",
    "revenue_uplift_proxy_pct"
]])

# Plot 1: Histogram of uplift distribution
plt.figure(figsize=(10, 6))
plt.hist(mc_df["revenue_uplift_proxy_pct"], bins=30)
plt.xlabel("Revenue uplift proxy %")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Revenue Uplift")
plt.show()

# Plot 2: Elasticity vs uplift
plt.figure(figsize=(10, 6))
plt.scatter(mc_df["elasticity"], mc_df["revenue_uplift_proxy_pct"], alpha=0.5)
plt.xlabel("Elasticity")
plt.ylabel("Revenue uplift proxy %")
plt.title("Elasticity vs Revenue Uplift")
plt.show()

# Plot 3: Average price change vs uplift
plt.figure(figsize=(10, 6))
plt.scatter(mc_df["avg_price_change_pct"], mc_df["revenue_uplift_proxy_pct"], alpha=0.5)
plt.xlabel("Average price change %")
plt.ylabel("Revenue uplift proxy %")
plt.title("Average Price Change vs Revenue Uplift")
plt.show()
```
